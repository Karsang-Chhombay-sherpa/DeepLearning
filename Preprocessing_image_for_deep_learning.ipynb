```python
#converting RGB images into Grayscale image using OpenCV library
import cv2
img=cv2.imread('/content/Dog.jpg')
```

```python
img.shape
```

```python
type(img)
```

```python
grayscale_img=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
grayscale_img.shape
```

```python
# sv2,imshow().will display the image . but not allowed in google colab

from google.colab.patches import cv2_imshow
cv2_imshow(grayscale_img)
```

```python
#saving the grayscale image
cv2.imwrite('dog_grayscale_image.jpg',grayscale_img)
```
